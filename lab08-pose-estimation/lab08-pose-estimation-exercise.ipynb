In this exercise, you will explore **human pose estimation** by running two different state-of-the-art models: **YOLOv7 Pose** and **OpenPose**. You will apply both models to video data and compare the output.

The goal is to understand how different pose estimation approaches work in practice and how model design affects performance in real-world scenarios.

### YOLOv7 Pose Estimation

**YOLOv7 Pose** extends the YOLO (You Only Look Once) object detection framework to perform **real-time multi-person pose estimation**.  

Key characteristics:
- Uses a **single-stage architecture** (detects people and keypoints in one pass).
- Fast and suitable for **real-time applications**.
- Outputs bounding boxes along with **keypoint coordinates**.
- Efficient for videos with multiple people.

Unlike traditional pipelines that first detect people and then estimate poses separately, YOLOv7 integrates both steps into one unified model, improving speed and efficiency.

````python
from pathlib import Path

YOLO_DIR = Path("/content/arti560-computer-vision-labs/lab08-pose-estimation/yolov7")
exercise_script = YOLO_DIR / "yolov7-pose-exercise.py"

code = r'''
import time
from pathlib import Path

import cv2
import numpy as np
import torch
from torchvision import transforms

from utils.datasets import letterbox
from utils.general import non_max_suppression_kpt
from utils.plots import output_to_keypoint, plot_skeleton_kpts


# ------------------------- SETTINGS -------------------------
input_size = 256

VIDEO_PATH = "/content/arti560-computer-vision-labs/lab08-pose-estimation/media/walking-persons.mp4"
OUTPUT_PATH = "/content/arti560-computer-vision-labs/lab08-pose-estimation/yolov7/yolov7_pose_walking_persons.avi"
WEIGHTS_PATH = "yolov7-w6-pose.pt"
# ------------------------------------------------------------


def pose_video(frame):
    img = letterbox(frame, input_size, stride=64, auto=True)[0]

    img = transforms.ToTensor()(img)
    img = torch.tensor(np.array([img.numpy()]))
    img = img.to(device)

    with torch.no_grad():
        t1 = time.time()
        output, _ = model(img)
        t2 = time.time()

    fps = 1 / (t2 - t1)

    output = non_max_suppression_kpt(
        output,
        0.25,
        0.65,
        nc=1,
        nkpt=17,
        kpt_label=True
    )

    output = output_to_keypoint(output)

    nimg = img[0].permute(1, 2, 0) * 255
    nimg = nimg.cpu().numpy().astype(np.uint8)
    nimg = cv2.cvtColor(nimg, cv2.COLOR_RGB2BGR)

    for idx in range(output.shape[0]):
        plot_skeleton_kpts(nimg, output[idx, 7:].T, 3)

    return nimg, fps


if __name__ == "__main__":
    device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
    print("Selected Device:", device)

    if not Path(WEIGHTS_PATH).exists():
        print("Downloading YOLOv7 pose weights...")
        import urllib.request
        url = "https://github.com/WongKinYiu/yolov7/releases/download/v0.1/yolov7-w6-pose.pt"
        urllib.request.urlretrieve(url, WEIGHTS_PATH)
        print("Weights downloaded.")

    print("Loading model...")
    weights = torch.load(WEIGHTS_PATH, map_location=torch.device("cpu"), weights_only=False)
    model = weights["model"]
    model.float().eval()
    model.to(device)

    cap = cv2.VideoCapture(VIDEO_PATH)

    if not cap.isOpened():
        raise FileNotFoundError(f"Could not open video: {VIDEO_PATH}")

    fps_in = cap.get(cv2.CAP_PROP_FPS)
    if fps_in <= 0:
        fps_in = 10

    ret, frame = cap.read()
    if not ret:
        raise RuntimeError("Could not read the first frame from the video.")

    processed_frame, _ = pose_video(frame)
    out_h, out_w = processed_frame.shape[:2]

    out = cv2.VideoWriter(
        OUTPUT_PATH,
        cv2.VideoWriter_fourcc(*"XVID"),
        fps_in,
        (out_w, out_h)
    )

    frame_count = 0
    start_time = time.time()

    while ret:
        processed_frame, fps_model = pose_video(frame)

        cv2.putText(
            processed_frame,
            "YOLOv7 Pose",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2,
            cv2.LINE_AA
        )

        cv2.putText(
            processed_frame,
            f"FPS: {fps_model:.2f}",
            (20, 80),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2,
            cv2.LINE_AA
        )

        out.write(processed_frame)
        frame_count += 1

        if frame_count % 20 == 0:
            print(f"Processed {frame_count} frames...")

        ret, frame = cap.read()

    cap.release()
    out.release()

    total_time = time.time() - start_time
    print("YOLOv7 Pose completed successfully.")
    print("Frames processed:", frame_count)
    print("Total time:", round(total_time, 2), "seconds")
    print("Output saved to:", OUTPUT_PATH)
'''

exercise_script.write_text(code, encoding="utf-8")

print("Fixed script created:")
print(exercise_script)
````

````python
from pathlib import Path

LAB_DIR = Path("/content/arti560-computer-vision-labs/lab08-pose-estimation")
POSE_DIR = LAB_DIR / "pose"

openpose_script = POSE_DIR / "OpenPoseVideo_exercise.py"

code = r'''
import cv2
import time
import numpy as np
from pathlib import Path

# ------------------------- SETTINGS -------------------------
VIDEO_PATH = "/content/arti560-computer-vision-labs/lab08-pose-estimation/media/walking-persons.mp4"
OUTPUT_PATH = "/content/arti560-computer-vision-labs/lab08-pose-estimation/pose/openpose_walking_persons.mp4"

MODE = "MPI"
PROTO_FILE = "/content/arti560-computer-vision-labs/lab08-pose-estimation/pose/mpi/pose_deploy_linevec_faster_4_stages.prototxt"
WEIGHTS_FILE = "/content/arti560-computer-vision-labs/lab08-pose-estimation/pose/mpi/pose_iter_160000.caffemodel"

inWidth = 368
inHeight = 368
threshold = 0.1

nPoints = 15
POSE_PAIRS = [
    [0,1], [1,2], [2,3], [3,4],
    [1,5], [5,6], [6,7],
    [1,14], [14,8], [8,9], [9,10],
    [14,11], [11,12], [12,13]
]
# ------------------------------------------------------------


print("OpenPose mode:", MODE)
print("Video:", VIDEO_PATH)
print("Output:", OUTPUT_PATH)

if not Path(PROTO_FILE).exists():
    raise FileNotFoundError(f"Missing prototxt file: {PROTO_FILE}")

if not Path(WEIGHTS_FILE).exists():
    raise FileNotFoundError(f"Missing model file: {WEIGHTS_FILE}")

cap = cv2.VideoCapture(VIDEO_PATH)

if not cap.isOpened():
    raise FileNotFoundError(f"Could not open video: {VIDEO_PATH}")

fps_in = cap.get(cv2.CAP_PROP_FPS)
if fps_in <= 0:
    fps_in = 10

ret, frame = cap.read()
if not ret:
    raise RuntimeError("Could not read first frame.")

frameHeight, frameWidth = frame.shape[:2]

writer = cv2.VideoWriter(
    OUTPUT_PATH,
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps_in,
    (frameWidth, frameHeight)
)

net = cv2.dnn.readNetFromCaffe(PROTO_FILE, WEIGHTS_FILE)
net.setPreferableBackend(cv2.dnn.DNN_BACKEND_OPENCV)
net.setPreferableTarget(cv2.dnn.DNN_TARGET_CPU)

print("Using OpenCV DNN CPU for OpenPose.")

frame_count = 0
start_time = time.time()

while ret:
    t = time.time()

    inpBlob = cv2.dnn.blobFromImage(
        frame,
        1.0 / 255,
        (inWidth, inHeight),
        (0, 0, 0),
        swapRB=False,
        crop=False
    )

    net.setInput(inpBlob)
    output = net.forward()

    H = output.shape[2]
    W = output.shape[3]

    points = []

    for i in range(nPoints):
        probMap = output[0, i, :, :]
        _, prob, _, point = cv2.minMaxLoc(probMap)

        x = int((frameWidth * point[0]) / W)
        y = int((frameHeight * point[1]) / H)

        if prob > threshold:
            points.append((x, y))
        else:
            points.append(None)

    for pair in POSE_PAIRS:
        partA, partB = pair

        if points[partA] and points[partB]:
            cv2.line(frame, points[partA], points[partB], (0, 255, 255), 3, lineType=cv2.LINE_AA)
            cv2.circle(frame, points[partA], 6, (0, 0, 255), thickness=-1, lineType=cv2.FILLED)
            cv2.circle(frame, points[partB], 6, (0, 0, 255), thickness=-1, lineType=cv2.FILLED)

    cv2.putText(
        frame,
        "OpenPose",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 50, 0),
        2,
        lineType=cv2.LINE_AA
    )

    cv2.putText(
        frame,
        "time taken = {:.2f} sec".format(time.time() - t),
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 50, 0),
        2,
        lineType=cv2.LINE_AA
    )

    writer.write(frame)
    frame_count += 1

    if frame_count % 20 == 0:
        print(f"Processed {frame_count} frames...")

    ret, frame = cap.read()

cap.release()
writer.release()

total_time = time.time() - start_time

print("OpenPose completed successfully.")
print("Frames processed:", frame_count)
print("Total time:", round(total_time, 2), "seconds")
print("Output saved to:", OUTPUT_PATH)
'''

openpose_script.write_text(code, encoding="utf-8")

print("Fixed OpenPose script created:")
print(openpose_script)
````

### Comparison between YOLOv7 Pose and OpenPose

Both YOLOv7 Pose and OpenPose were applied to the same video, `walking-persons.mp4`, from the media folder. YOLOv7 Pose completed the video faster because it performs person detection and keypoint estimation efficiently in one deep learning pipeline. OpenPose was slower because it estimates body keypoints using a more detailed pose estimation process.

In terms of output, YOLOv7 Pose was more suitable for real-time video processing, while OpenPose helped visualize the classical pose-estimation idea of detecting body joints and connecting them into a human skeleton. Overall, YOLOv7 Pose is faster, while OpenPose is useful for understanding the keypoint-based skeleton structure.
